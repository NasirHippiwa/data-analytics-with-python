```python
class Person:
    def __init__(self,name, age, nic):
        self.name = name
        self.age = age
        self.nic = nic 
    def display_inf(self):
        print(f"Role : {self.role}")
        print (f"Name : {self.name} , Age : {self.age} , NIC.No : {self.nic}")
        return " "

    def rol(self):
        return "Person"


class Adim(Person):
    def __init__(self,name,age,nic, department, salary):
        super().__init__(name,age,nic)
        self.department = department
        self.salary = salary

    def display_inf(self):
        super().display_inf()
        print("Department : ", self.department)
        print("Salary : ", self.salary)

a = Adim("Ali", 21, 7130154884591,"HR",50000)

print(a.display_inf())
print(a.rol())
```
