```python
# 3 write a program in which use all the operators we can use in python

a = 10
b = 5

# 1 Arithmetic operators
print("addition:", a + b)
print("subtraction:", a - b)
print("multiplicaion:", a * b)
print("division:", a / b)
print("modulus:", a % b)
print("exponentiation:", a ** b)
print("floor division:", a // b)
```

```python
# 2 Assignment operators
c = 10 # assignment with =
c += 5  # equivalent to c = c + 5
print("c after addition:", c)

c -= 3  # equivalent to c = c - 3
print("c after subtraction:", c)

c *= 2  # equivalent to c = c * 2
print("c after multiplication:", c)

c /= 4  # equivalent to c = c / 4
print("c after division:", c)
```

```python
# 3 comparison operators
x = 10
y = 5
print("x is equal to y:", x == y)
print("x is not equal to y:", x != y)
print("x is greater than y:", x > y)
print("x is less than y:", x < y)
print("x is greater than or equal to y:", x >= y)
print("x is less than or equal to y:", x <= y)
```

```python
# 4 Bitwise Operators
x = 10
y = 4
print("Bitwise AND:", x & y)
print("Bitwise OR:", x | y)
print("Bitwise XOR:", x ^ y)
print("Bitwise NOT:", ~x)
print("Left Shift:", x << 1)
print("Right Shift:", x >> 1)
```

```python
# 5 Logical Operators
a = 10
b = 5
x = True  
y = False
print("Logical AND:", x and y)
print("Logical OR:", x or y)
print("Logical NOT:", not x)
print("AND:", a > 5 and b < 5)
print("OR:", a > 20 or b < 5)
print("NOT:", not(a > b))
```

```python
# 4 complete the following steps of small task:
# 1. Mention Marks OF English, Islamiat and Maths out of 100 in 3 different variables
# 2. mention Variable of Total Marks and assign 300 to it.
# 3. Calculate Percentage.
English = 68
Islamiat = 73
Maths = 92

Total_marks = 300

percentage = (English + Islamiat + Maths) / Total_marks * 100

print("Englist mark = ",English)
print("Islamiat mark = ",Islamiat)
print("Math mark = ", Maths)
print("Total Marks = ", Total_marks)
print("Percentage:", percentage, "%")
```

```python
# 6 Take the radius of a circle from a variable and calculate its Area and Circumference(use pi = 3.14159).

r = 25
pi = 3.14159

area = pi * r ** 2
# area = are * 2
circumference = 2 * r *pi

print("Area of Circle:",area)
print("Circumference of Circle:",circumference)
```
