```python
# Task -- calculate grade of student s

# 90+ A+

# 80 - 90 -- A

# 70 - 80 B

# 60 - 70 - C

# define a function get_grade




def get_grade(score):
    if score >= 90:
        return "A+"
    elif score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "F"

students = [

    {"name": "Ali", "score": 92},

    {"name": "Sara", "score": 75},

    {"name": "Ahmed", "score": 55},

    {"name": "Zara", "score": 88}

]


for i in students:
    name = i["name"]
    score = i["score"]
    grade = get_grade(score)
    print(f"{name} scored {score} and received grade {grade}.")
```

```python
def total_amount(food_amount, discount=0 , tax_rate=15):
    discount_amount = food_amount * discount / 100
    food_amount -= discount_amount
    tax = food_amount * tax_rate / 100
    total = food_amount + tax
    return total

total_amount(100, 10, 8)
```

### write a function that take a parameter and check if the number is prime or not.

```python
def check_prime(num):
    if num == 1:
        print(f"{num} is not a prime number.")
    elif num == 2:
        print(f"{num} is a prime number.")
    else:
        for i in range(2, num ):
            if num % i == 0:
                print(f"{num} is not a prime number.")
                break
        else:
            print(f"{num} is a prime number.")

check_prime(0)
```

### write a recursion function to sum of the list 

```python
list = [1, 2, 3, 4, 5]

def sum_of_list(lst):
   if len(lst) == 1:
       return lst[0]
   else:
       return lst[0] + sum_of_list(lst[1:])


sum_of_list(list)
```

### Write a program to solve the fibonacci Sequence 
### Using recursion function

```python
def fs(num):
    if num ==1:
        return 0
    elif num == 2:
        return 1
    else:
        return fs(num - 1) + fs(num - 2)

print(fs(7))
```
