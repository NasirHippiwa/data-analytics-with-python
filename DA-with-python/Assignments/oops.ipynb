```python
class Student:

    def __init__(self,name,age,math,phy,chy):
        self.name = name
        self.age = age
        self.math = math
        self.phy = phy
        self.chy = chy
        self.total = (math + phy + chy)
        self.score = self.total /300
        self.percentage = self.score * 100

    def score1(self):
       if self.percentage >= 90:
            return "A Grade"
       elif self.percentage >= 80:
            return "B Grade"
       else:
            return "C Grade"


s1 = Student("Yasir", 19,82,79,91)

print(s1.name)
print(s1.age)
print(s1.total)
print(s1.percentage)
print(s1.score1())

```

```python
class Person:                                 # Parent Class

    def __init__(self, name, age):
        self.name = name
        self.age  = age

    def greet(self):
        return(f"Hi, I am {self.name}")
    
    def show_age(self):
        return(f"I am {self.age} years old")

class Student(Person):

    def study(self):
        return(f"{self.name} is studying! ")

class Teacher(Person):

    def teach(self):
        return(f"{self.name} is teaching! ")


student = Student("nasir",24)

student_1 = Student("Yasir", 21)

print(student_1.name)

print(student_1.greet())

print(student_1.show_age())

print(student_1.study())

teacher1 = Teacher("Ayan", 25)

print(teacher1.name)

print(teacher1.greet())

print(teacher1.show_age())

print(teacher1.teach())
```
