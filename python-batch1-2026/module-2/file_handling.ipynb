
# class activity

 write a program that

 ask user to enter a message

 appends it to log.txt with the entry number

 keeps asking user until user types "quite"

 at the end, read and prints the messages from the files from log.txt

```python
open("log.txt", "w").close  # Create or clear the log file
entry_no = 1
while True:
    user_input = input("Enter a line to append to the file (or type 'quit' to quit): ")
    if user_input.lower() == 'quit':
        break
    else:
        with open("log.txt", "a") as file:
            file.write(f"{entry_no}: {user_input}\n")
        entry_no += 1


with open("log.txt", "r") as file:
    content = file.read()
    # print("\nLog File Content:")
    print(content)
```

```python
import csv
eligibles = []
not_eligibles =[]
with open("data.csv", "r") as file:
    readers = csv.DictReader(file)
    for row in readers:
        if row["age"] == "age":
            continue

        if int(row["age"]) >= 18:
            eligibles.append(f'{row["id"]} {row["name"]} S/O {row["fname"]} from {row["city"]} is eligible for this course')
        else:
            not_eligibles.append(f'{row["id"]} {row["name"]} S/O {row["fname"]} from {row["city"]} is not eligible for this course')
print("Eligibles")      
for i in eligibles: print(i)
print("Not Eligibles")
for i in not_eligibles: print(i)
```

```python
blance = 100
def cash_withdraw(amount):
    global blance
    if amount <0:

        raise ValueError("Amount cannot be negative")

    elif amount > blance:
        raise ValueError("Not enough balance")



print(cash_withdraw(100))

# print(load_record(record, "email"))
```

```python
balance = 100

def withdraw(amount):

    if amount < 0:

        raise ValueError("Amount cannot be negative")

   

    if amount > balance:

        raise ValueError("Not enough balance")

   

    new_balance = balance - amount

    return new_balance

try:

    result = withdraw(50)

    print("Withdraw successful, remaining balance:", result)

except ValueError as e:

    print("Error:", e)

try:

    result = withdraw(-20)

    print("Withdraw successful, remaining balance:", result)

except ValueError as e:

    print("Error:", e)

try:

    result = withdraw(200)

    print("Withdraw successful, remaining balance:", result)

except ValueError as e:

    print("Error:", e)
```
