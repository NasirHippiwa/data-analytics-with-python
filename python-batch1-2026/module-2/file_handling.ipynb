
# class activity

 write a program that

 ask user to enter a message

 appends it to log.txt with the entry number

 keeps asking user until user types "quite"

 at the end, read and prints the messages from the files from log.txt

```python
open("log.txt", "w").close  # Create or clear the log file
entry_no = 1
while True:
    user_input = input("Enter a line to append to the file (or type 'quit' to quit): ")
    if user_input.lower() == 'quit':
        break
    else:
        with open("log.txt", "a") as file:
            file.write(f"{entry_no}: {user_input}\n")
        entry_no += 1


with open("log.txt", "r") as file:
    content = file.read()
    # print("\nLog File Content:")
    print(content)
```

```python
open("data.xlsx", "w").close  # Create or clear the Excel file

with open("data.xlsx", "a") as file:  # Open the Excel file in append mode
    file.write("Name, Age, City\n")
    file.write("Alice, 30, New York\n")
    file.write("Bob, 25, Los Angeles\n")
    file.write("Charlie, 35, Chicago\n")
```

```python
import openpyxl

workbook = openpyxl.Workbook()
sheet = workbook.active
sheet["A1"] = "Name"
sheet["B1"] = "Age"
sheet["C1"] = "City"
sheet["A2"] = "Alice"
sheet["B2"] = 30
sheet["C2"] = "New York"
sheet["A3"] = "Bob"
sheet["B3"] = 25
sheet["C3"] = "Los Angeles"
sheet["A4"] = "Charlie"
sheet["B4"] = 35
sheet["C4"] = "Chicago"

workbook.save("data.xlsx")
```

```python
import openpyxl

workbook = openpyxl.load_workbook("data.xlsx")
sheet = workbook.active
for row in sheet.iter_rows(values_only=True):
    print(row)
```
